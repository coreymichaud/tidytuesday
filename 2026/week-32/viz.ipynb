# Data Wrangling

# Styling

```python
# Custom theme
my_theme = go.layout.Template(layout=go.Layout(paper_bgcolor="white"))

fig.update_layout(template=my_theme);
```

# Layout

```python
# Figure layout
fig.update_layout(
    width=600,
    height=600,
    xaxis=dict(title="Nitrogen strength relative to hydrogen →"),
    yaxis=dict(title="Oxygen strength relative to hydrogen →"),
    margin=dict(t=100, b=100),
    legend=dict(title=dict(text="Activity Type")),
);
```

# Title & Subtitle

```python
# Subtitle
my_subtitle = dict(
    text=wrap_subtitle("Hopefully I don't need a longer subtitle...", 50),
    font=dict(size=12, color="grey"),
)

# Title
my_title = dict(
    text="BPT Diagnosis Diagram",
    subtitle=my_subtitle,
    x=0.5,
    xref="paper",
    xanchor="center",
)

fig.update_layout(title=my_title);
```

# Footer

```python
# Footer
footer = [
    dict(
        # text="<i><b>Source:</b> Ho, Filippenko & Sargent (1995, 1997, 2009) via VizieR.<br><b>Author:</b> Corey Michaud for #TidyTuesday Y26W32.</i>",
        text="<i><b>Source:</b> Ho, Filippenko & Sargent (1995, 1997, 2009) via VizieR. | <b>Author:</b> Corey Michaud for #TidyTuesday Y26W32.</i>",
        xref="paper",
        yref="paper",
        x=0.5,
        y=-0.18,
        xanchor="center",
        yanchor="top",
        showarrow=False,
        font=dict(size=8, color="gray"),
    )
]


fig.update_layout(annotations=footer);
```

# Ta-da!

```python
# Show figure, with configurations
fig.show(config={"staticPlot": True})

# Save figure as a PNG
fig.write_image("viz.png", scale=2)
```
